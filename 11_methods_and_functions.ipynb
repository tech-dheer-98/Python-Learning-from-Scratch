```python
# def with return keyword instead of print, bcoz return can store in a var 

def add_num(num1,num2):
    num=num1+num2
    return f"The sum of {num1} and {num2} is {num}"

add_num(40,58)
```

```python
# Difference between return and print in def function 

def print_result(a,b):
    print(a+b)

def return_result(a,b):
    return a+b
```

```python
# Logic with Python Functions 
# Check for Even number 

def even_check(number):
    return number%2==0

even_check(21)
```

```python
# Return True if any number is even inside a list 

l=[x for  x in range(1,11,2)]

def even_check(number):
    for num in number:
        if num%2==0:
            return True
        else :
            pass
    return False

#Calling a Function
even_check(l)
```

```python
# Return Even Numbers in a list 

def return_even_list(number):

    #Placeholder variables
    even_numbers=[]

    for num in number:
        if num%2==0:
            even_numbers.append(num)
        else:
            pass
    return even_numbers

l=[x for  x in range(0,11,1)]
#Calling a Function
return_even_list(l)

```

```python
# Tuples Unpacking with Python Functions 

# Check who has the highest working hours among the list 

work_hours=[('Dheer',115),('Kartik',72),('Gopal',55)]


def employee_work_hours(work_hours):
    current_max=0
    employee_of_month=''

    for employee, hours in work_hours:
        if hours>current_max:
            current_max=hours
            employee_of_month=employee
        else:
            pass

    return (employee_of_month,current_max)



# Calling the Function 
employee_work_hours(work_hours)
```

```python
# Tuple Unpacking 
name,hours=employee_work_hours(work_hours)
```
