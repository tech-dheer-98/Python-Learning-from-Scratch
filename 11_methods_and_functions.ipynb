```python
# Return True if any number is even inside a list 

l=[x for  x in range(1,11,2)]

def even_check(number):
    for num in number:
        if num%2==0:
            return True
        else :
            pass
    return False

#Calling a Function
even_check(l)
```

```python
# Return Even Numbers in a list 

def return_even_list(number):

    #Placeholder variables
    even_numbers=[]

    for num in number:
        if num%2==0:
            even_numbers.append(num)
        else:
            pass
    return even_numbers

l=[x for  x in range(0,11,1)]
#Calling a Function
return_even_list(l)

```

```python
# Tuples Unpacking with Python Functions 

# Check who has the highest working hours among the list 

work_hours=[('Dheer',115),('Kartik',72),('Gopal',55)]


def employee_work_hours(work_hours):
    current_max=0
    employee_of_month=''

    for employee, hours in work_hours:
        if hours>current_max:
            current_max=hours
            employee_of_month=employee
        else:
            pass

    return (employee_of_month,current_max)



# Calling the Function 
employee_work_hours(work_hours)
```

```python
# Tuple Unpacking 
name,hours=employee_work_hours(work_hours)
```

```python
# *args and **kwargs

# *args 
def myfunc(*args):
    for item in args:
        print(item)
    

#calling the *args function,it treats as a tuple
a=myfunc(10,"Dheer","Sneha","Sushmita","Amrine")

#It is not stored 
type(a)
```

```python
# **kwargs

def myfunc(**kwargs):
    if 'fruit' in kwargs:
        print(kwargs)
        print(f"My fruit of choise is {kwargs['fruit']}.")
    else:
        print("I did not find any fruit here ")

myfunc(veggies='lattice',brekfast='eggs',fruit='apple')
```

```python
# Using *args and **kwargs in combination :

def myfunc(*args,**kwargs):
    print(args)
    print(kwargs)
    return (f'I would like to have {args[1]} {kwargs['gender']}')


myfunc(10,20,30,food='eggs',animal='lion',gender='female')
```

```python
# Define a function called myfunc that takes in an arbitrary number of arguments,
# and returns a list containing only those arguments that are even.

def myfunc(*args):
    a=[x for x in args if x%2==0]
    return a

myfunc(1,2,3,4,7,8,9,12)
```

```python
#Define a function called myfunc that takes in a string,
#and returns a matching string where every even letter is uppercase,
#and every odd letter is lowercase. 
#Assume that the incoming string only contains letters, 
#and don't worry about numbers, spaces or punctuation. 
#The output string can start with either an uppercase or lowercase letter, 
#so long as letters alternate throughout the string.


def myfunc(word):
    result=''
    for i,char in enumerate(word):
        if i%2==0:
            result+=char.upper()
        else:
            result+=char.lower()
            
    return result

# Calling the Function
myfunc("Dheer")
```
