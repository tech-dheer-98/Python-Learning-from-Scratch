# While Loop : 

```python
# Syntax :

'''
while "some_bool_condition":
        # do something
        

# While combined with and else statement 

while " some_bool_condition":
    # do something
    
else:
    # do something different
'''
```

```python
# While Loop

x=0


while x<5:
    print(f"The Current value of x is {x}")
    # If we dont add this line , it will run a infinite while loop 
    x+=1
else:
    print(f"x is not less than {x}")

```

## Break,Continue,pass
- We can use break,continue, pass statements in our loops to add additional functionality for various 
  cases . The three statements are defined by 

1) break : Breaks out of the current closest enclosing loop. 
2) continue: Goes to the top of the closest enclosing loop.
3) pass : Does nothing at all 


```python
# pass : does nothing at all
x=[1,2,3,4]

for items in x:
    pass

print("I like the way you code")
```

```python
# Continue : Skips the letter wherever found   and continues ahead 
for i in "Dheer Dalal":
    if i=="D":
        continue
    print(i)
```

```python
# break : Wherever finds the letter breaks out of the loop 
for i in "Dheer Dalal":
    if i=="r":
        break
    print(i)
    
```

```python
x=0

while x<5:
    if x==2:
        break
    print(x)
    x+=1
    
```
