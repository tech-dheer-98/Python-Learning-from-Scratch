##  Basic Data Types 

1) int   | Whole numbers such as : 3,10,12
2) float | Numbers with Decimal point : 2.3,3.5,5.5
3) str   | Ordered sequence of characters : "hello","Samay",2000,5000

## Data Structures 
1) list  | Ordered sequence of objects : [10,"Helo","Student",45.3]
2) dict  | Unordered Key: Value pairs :{"key_1":"value_1","name":"frankie"}
3) tup   | Ordered immutable sequence of objects (10,"Helo","Student",45.3)
4) set   | Unordered collection of unique objects : {"a","b","c"}
8) bool  | Logical Value Indicating True or False
