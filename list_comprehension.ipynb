# List Comprehension

```python
# String  to list Comphrension
mylist=[x for x in 'Dheer']  # 
mylist
```

```python
# Range to List Comprehnsion to print square of numbers in a range  

mylist=[num**2 for num in range(0,6)]
mylist
```

```python
# Want to Print only even numbers and Square them  using list Comphrension

mylist=[ x**2 for x in range(0,10) if x%2==0]
mylist
```

```python
# Convert Temp to Fahrenhit using List Comphrension
celcius=[32.5,69.45,55.1]

fah= [((9/5)*temp+32) for temp in celcius]
fah
```

```python
# Traditional for-loop method 
celcius=[32.5,69.45,55.1]
fah=[]

for temp in celcius:
    fah.append(((9/5)*temp+32))
print(fah)
```

```python
# Using If-else statement in List Comphrension 

for x in range(0,10):
    if x%2==0:
        print((x))
    else:
        print("Odd")
```

```python
# If-else List Comprehension

results = [x if x%2==0 else 'Odd' for x in range (0,11)]
results
```

```python
# Nested Loops Using For-Loop

mylist=[]

for x in [2,4,6]:
    for y in [100,200,500]:
        mylist.append(x*y)
mylist
```

```python
# Nested Loops Using List Comprehension 

mylist=[x*y for x in [2,4,6] for y in [100,200,500]]
mylist
```
