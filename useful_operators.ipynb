# Useful Operators 
1) Range 
2) List with Range Function
3) Enumerate Function
4) Zip
5) IN Operator 
6) Min,Max 
7) Random Library(shuffle,randint)
8) User Input Accceptance (input)

```python
#Range 

# Step size for range function 
for num in range(0,11,2):
    print(num)
```

```python
# Range with List 

a=list(range(1,20,2))
a
```

```python
# Enumerate Fucntion 
bag=[("Dheer",5),("Anita",6),("Anu",7)]

#This gives index position of each element 
for item in enumerate(bag):
    print(item)

```

```python
# Zip Function : Tuple Unpacking 

list1=(1,2,3,4)
list2=(87,88,90,"Dheer")
list3=["Annu","Hemali","Krina","Anjali"]

for item in zip(list1,list2,list3):
    print(item)
```

```python
# We want to unpack as list 
a=list(zip(list1,list2,list3))
a[1]
```

```python
# In Operator using List 

'x' in ['x','y','z']
```

```python
# In Operator using String 

'h' in 'Dheer Dalal'
```

```python
# In Operator in Dictionary 

#Keys 
'key_1' in {'key_1':1,'key_2':2,'key_1':11}
```

```python
#Values

d={'key_1':1,'key_2':2,'key_1':11}

2 in d.values()
```

```python
# Min 

list_new=[10,20,30,70,440,11,6]

min(list_new)
```

```python
# Max

max(list_new)

```

```python
# random library 

from random import shuffle

shuffle(list_new)
list_new
```

```python
# Shuffle does not store anything 
type(shuffle(list_new))
```

```python
from random import randint

# Random Integer will be displayed from the given Range 

rand_num=randint(1,30)
rand_num
```

```python
# User Input 

mob=int(input("Enter your phone number"))
print(mob,type(mob))
```
