# Reading ,Writing, Appending Modes
- mode ="r"  is read only 
- mode ="w" is write only ( will overwrite files or create new!)
- mode ="a" is append only (will add on to files )
- mode ="r+" is reading and writing
- mode ="w+" is writing and reading (Overwrites exixting files or creates a new file )

```python
# Open a file in current path  :

myfile=open('temp.txt')

# This will read as a string
myfile.read()
```

```python
# To reset the cursor 
myfile.seek(0)

# This will read the file line by line as a list
n=myfile.readlines()
n
```

```python
# Technical way to open and close any file from a different path 

# Read Mode 
with open("D:\\DHEER USERS\\Desktop\\API Development\\fastapi.txt",mode="r") as d:
    print(d.read())

```

```python
# Append new line 
with open ("temp.txt",mode="a") as f:
    f.write("\nThis is the Append method in code \nThis is the new line of append ")
```

```python
#Read the appended  line 
with open ("temp.txt",mode="r") as d:
    print(d.read())
```

```python
# Overwriting the file adn Reading it 
with open ("unnss.txt",mode="w") as u:
    u.write("How are you\nWho is your Daddy ")

with open("unnss.txt",mode="r") as f:
    print(f.read())
```
