# For Loops :


```python
# Syntax

my_list=[1,2,3,5,8,7,96,55]

for num in my_list:
    print(num)
```

```python
# To print only even numbers in my_list 

for num in my_list:
    # Check for even
    if num%2==0:
        print(num)
    else :
        pass
```

```python
# ForLoop with List 


# Sum the numbers in a list 
my_list=[1,2,3,4]

#Initialize
sum_list=0

for num in my_list:
    sum_list=sum_list+ num
print(sum_list)
```

```python
# For loop with String 

mystring="Who is your Daddy ?"

for letter in mystring:
    print(letter)
```

```python
# ForLoop in Tuple

tup=(1,"Dheer","Rashmi")

for x in tup:
    print(x)
```

```python
# Tuple Unpacking 

my_list=[(1,3),(5,6),(8,9)]

for a,b in my_list :
    print(a)
    print(b)
```

```python
my_list=[(1,2,5),(6,8,9),(67,66,98)]

for a,b,c in my_list :
    print(a)
    print(b)
```

```python
# ForLoop in Dictionaries:

d={"k1":1,"k2":2,"k3":3}

# By default we iterate only keys 
for x in d :
    print(x)

# But if we want to iterate keys and values both 

for y in d.items():
    print(y)

# Suppose we want to iterate only key or only value 

for k,v in d.items():
    print(f"Key:{k}")
    print(f"Value:{v}")
```
