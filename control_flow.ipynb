# Control Flow Logic :
1) if 
2) elif
3) else 


```python
# If -else Statement  
hungry =True

if hungry:
    print("Feed Me !") 
else:
    print("I am not Hungry")
```

```python
# If-elif-else Statement

km=10001

if km<5000:
    print("You need to run more !")

elif km==5000:
    print("More 250 Calories to Burn ")


elif km==10000:
    print("Congrats you have completed 500 cals burning !!!")

else:
    print("You walked more than 10k Steps")
 
```
