## Lambda Expressions ,Maps and Filter :

- They are ananomous functions or one time used fucntion 

# Map Function 

```python
#Defining a Squre function
def square(num):
    return num**2

#List 
my_num=[x for x in range(1,11)]

# Using the Map Fucntion 
list(map(square,my_num))
```

```python
#Defining a Slicer Function 

def splicer(mystring):
    if len(mystring)%2==0:
        return 'Even'
    else:
        return len(mystring)
    
#List of names 
names=["Dheer","Mansi","Andy","Kartik","Anu"]

# Using the Map Function 
list(map(splicer,names)) 
```

# Filter Function 

```python
# Creating a Function 

def check_even(num):
    return num%2==0

list_num=[x for x in range(0,10)]

# Using the filter Function 

list(filter(check_even,list_num))
```

# Lambda Expression :


```python
square=lambda num:num**2

square(4)
```

```python
# Using the Lambda Function with Map 

list(map(lambda num:num**2,my_num))
```

```python
# Using the Lambda Function with Filter 

#Getting all the numbers from list those are odd
list(filter(lambda num:num%2!=0,list_num))
```
