```python
a="Dheer"
b="Anita"

# Indexing 
print(a[0])
print(a[-1])
```

```python
# Slicing 

my_string="abcdefghijk"
my_string[0:5]   # start : stop : step
```

```python
# Strings Properties and Method:

#Immutability
name ="Sam"

# We need to Concatenate String 
new_name=name[1:]
new_name="z"

x="X" + new_name
x*3

```

```python
# Upper,Lower,Split Function 
x='Hello Dheer Dalal how can I help you ?'

#Upper a String
upper=x.upper()
print(upper)

#Lower a String
lower=x.lower()
print(lower)

# Split a String 
split=x.split()
print(split)

# Split from a character where there is "o" in a variable

split_var=x.split("o")
split_var
```

# Method in List 
1) append() - Means to add an element in list at the end of list 
2) pop() - Means to remove last element from the list ,by default index -1
3) sort() - Sort the list in Ascending order or Alpabetical order
4) type() - Checks the type of the variable
5) reverse() - Reverses the list w/o sorting it 

```python
#append()
new_list.append('Dheer')
new_list
```

```python
#pop()
new_list.pop()
new_list
```

```python
# Want to pop from a specific index
new_list.pop(6)
new_list
```

```python
# Sort()
num_list=[4,45,3,96,77,22]
alpa_list=["Dheer","Rhea","Anjali","Gzel","Granny","Sesam","Patel"]
num_list.sort()
alpa_list.sort()
print("This is the Sorted Number list: {}".format(num_list))
print("This is the Sorted Alpha list: {}".format(alpa_list))
```

```python
# type()
type(num_list.sort()) # Means its not saving 
type(num_list)
```

```python
#reverse()
num_list=[4,45,3,96,77,22]
num_list.sort()
num_list.reverse()
print("This is the reverse of Num list :{}".format(num_list))
alpa_list.reverse()
print("This is the reverse of Alpha list :{}".format(alpa_list))

```

# Dictionaries 

```python
my_dict={'name':'Dheer','surname':'dalal'}

# Retreiving value by Key name 
print(f"The Value of the entered key is :{my_dict['name']}")
```

```python
# Dictionaries can also hold list

d={'k1':[1,'2',"swami ji",4,5,6],'k2':'dhananjay'}

#Retreving a value from the list and making it upper ,works only on string object
d['k1'][2].upper()
```

```python
# See all keys,values and keys-values

#dictinary 
print(f"Dictionary:{d}")

#keys of dictionary
print(f"Keys of Dictionary: {d.keys()} ")

#values of dictionary
print(f"Values of Dictionary: {d.values()}")

#items of dictionary 
print(f"Items of Dictionary: {d.items()} ")
```
